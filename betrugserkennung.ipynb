```python
##Betrugserkennung bei Kreditkartentransaktionen. Diese Projekt hilft die Betrug über Summe und Zeit von Transaktion richtig zu erkennen.

## 1. Daten laden
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/fraud-project/creditcard.csv')
print(df.shape)
print(df['Class'].value_counts())
```

```python
## 2. Explorative Datenanalyse
df.info()

## Keine Fehlende Werte. Jedes Feld besitzt ein Wert.
```

```python
df.describe()

## Aus diese Tabelle kann man folgende Daten auftufen:
## 1. Die Daten würden innerhalb von 48 Stunden gesammelt.
## 2. Amount liegt in Median bei 22 Euro pro Transaktion. Die Mittelwert liegt bei 88 Euro.
## 3. Der Mittelwert von Class zeigr das nur 0.17% sind Betrug.
```

```python
df['Class'].value_counts(normalize=True)

## Nur 0,17 % der Transaktionen sind Betrug
```

```python
df.duplicated().sum()
## Datensatz enthäjt 1081 doppelte Zeilen
```

```python
import matplotlib.pyplot as plt

df['Class'].value_counts().plot(kind='bar')
plt.title('Verteilung:normaler Transaktionen vs. Betrug')
plt.xlabel('Klasse 0-Normal, 1- Betrug')
plt.ylabel('Anzahl Transaktionen')
plt.show()
```

```python
df.groupby('Class')['Amount'].describe()

## Das sind ganz wichtige Daten, da man merkt hier besonders gut, dass Betrüger bezahlen in Median weniger, asl normaler Menschen, aber die Mittelwert bei Betrüger ist deutlich höhere.
```
