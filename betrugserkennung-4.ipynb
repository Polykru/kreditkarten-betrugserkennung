```python
## 3. Datenvorbereitung
df = df.drop_duplicates()
print(df.shape)
print(df['Class'].value_counts())
```

```python
X = df.drop('Class', axis=1)
Y = df['Class']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y , test_size=0.2, stratify=Y, random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train[['Time', 'Amount']]=scaler.fit_transform(X_train[['Time','Amount']])
X_test[['Time','Amount']]=scaler.transform(X_test[['Time','Amount']])
```

```python
Y_test.value_counts()

##Der Scaler lernt Mittelwert und Streuung nur aus den Trainingsdaten, damit keine Informationen aus den Testdaten ins Training gelangen – sonst wäre die Bewertung zu optimistisch (Data Leakage).
```

```python
from sklearn.linear_model import LogisticRegression

model_lr=LogisticRegression(max_iter=1000)
model_lr.fit(X_train,Y_train)
Y_pred_lr=model_lr.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(Y_test,Y_pred_lr))
print(classification_report(Y_test,Y_pred_lr))

##Die Logistic Regression erreicht eine Accuracy von nahezu 100 %, erkennt aber nur 59 % der Betrugsfälle (Recall).
##39 von 95 Betrugsfällen bleiben unentdeckt. Die Precision von 85 % zeigt, dass die meisten Alarme berechtigt sind – das Modell ist also eher vorsichtig.
##Accuracy ist hier als Metrik irreführend.
```

```python
from sklearn.ensemble import RandomForestClassifier

model_rf=RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42)
model_rf.fit(X_train, Y_train)
Y_pred_rf=model_rf.predict(X_test)

print(confusion_matrix(Y_test,Y_pred_rf))
print(classification_report(Y_test,Y_pred_rf))

##Der Random Forest übertrifft die Logistic Regression in beiden Kennzahlen:
##Er erkennt 73 % der Betrugsfälle (statt 59 %) bei einer Precision von 97 % (statt 85 %). Nur 2 ehrliche Kunden werden fälschlich blockiert.
##Dennoch bleiben 26 Betrugsfälle unentdeckt.
```

```python
Y_proba_rf=model_rf.predict_proba(X_test)[:,1]
print(Y_proba_rf[:50])
```

```python
from sklearn.metrics import precision_score, recall_score

for schwelle in [0.5,0.4,0.3,0.2,0.1]:
  Y_pred=(Y_proba_rf>=schwelle).astype(int)
  p=precision_score(Y_test, Y_pred)
  r=recall_score(Y_test, Y_pred)
  fp=((Y_pred==1)& (Y_test==0)).sum()
  print(f"Schwelle {schwelle}: Precision {p:.2f} | Recall {r:.2f} | Fehlalarme {fp}")

```

```python
from sklearn.metrics import PrecisionRecallDisplay

PrecisionRecallDisplay.from_predictions(Y_test, Y_proba_rf, name='Random Forest')
plt.title('Precision-Recall-Kurve')
plt.show()
```
